## Interpretation

The popularity baseline achieves modest performance.

The AUC score (0.5385) is slightly above random (0.5), indicating that global popularity provides some predictive signal for user clicks.

However, the relatively low MRR and nDCG values suggest that the model does not rank relevant news items effectively at top positions.
